```python
import matplotlib.pyplot as plt
from matplotlib.patches import FancyBboxPatch, FancyArrowPatch

# Function to draw a labeled box
def draw_box(ax, xy, text, box_color="#FFD966", text_color="black", width=2.5, height=1.0):
    x, y = xy
    bbox = FancyBboxPatch(
        (x - width/2, y - height/2),
        width, height,
        boxstyle="round,pad=0.3",
        fc=box_color,
        ec="black",
        lw=1.5
    )
    ax.add_patch(bbox)
    ax.text(x, y, text, ha="center", va="center", fontsize=11, weight='bold', color=text_color)

# Function to draw an arrow
def draw_arrow(ax, start, end, color="black"):
    arrow = FancyArrowPatch(start, end, arrowstyle='-|>', mutation_scale=20, color=color, lw=1.5)
    ax.add_patch(arrow)

# Create figure
fig, ax = plt.subplots(figsize=(6,5))
ax.set_xlim(-3, 3)
ax.set_ylim(-1, 5)
ax.axis('off')

# Draw boxes
draw_box(ax, (0,4), "BERT\n(768-dim)", box_color="#A9D0F5")
draw_box(ax, (0,2.5), "Dropout + FC1 + ReLU\n(256-dim)", box_color="#FFD966")
draw_box(ax, (0,1), "Dropout + FC2\n(5 classes)", box_color="#F5B7B1")

# Draw arrows
draw_arrow(ax, (0,3.5), (0,3))        # BERT → FC1
draw_arrow(ax, (0,2), (0,1.5))        # FC1 → FC2

# Title
ax.set_title("Single-Task BERT + MLP Model", fontsize=14, weight='bold')

plt.show()
```

```python
import matplotlib.pyplot as plt
from matplotlib.patches import FancyBboxPatch, FancyArrowPatch

# Function to draw a labeled box
def draw_box(ax, xy, text, box_color="#FFD966", text_color="black", width=2.5, height=1.0):
    x, y = xy
    bbox = FancyBboxPatch(
        (x - width/2, y - height/2),
        width, height,
        boxstyle="round,pad=0.3",
        fc=box_color,
        ec="black",
        lw=1.5
    )
    ax.add_patch(bbox)
    ax.text(x, y, text, ha="center", va="center", fontsize=11, weight='bold', color=text_color)

# Function to draw an arrow
def draw_arrow(ax, start, end, color="black"):
    arrow = FancyArrowPatch(start, end, arrowstyle='-|>', mutation_scale=20, color=color, lw=1.5)
    ax.add_patch(arrow)

# Create figure
fig, ax = plt.subplots(figsize=(8,5))
ax.set_xlim(-4, 4)
ax.set_ylim(-1, 5)
ax.axis('off')

# Draw boxes
draw_box(ax, (0,4), "BERT\n(768-dim)", box_color="#A9D0F5")
draw_box(ax, (0,2.5), "Shared Layer\nLinear + ReLU\n(256-dim)", box_color="#FFD966")
draw_box(ax, (-2,0.5), "Sentiment Head\n3 classes", box_color="#B5EAD7")
draw_box(ax, (2,0.5), "Star Rating Head\n5 classes", box_color="#F5B7B1")

# Draw arrows
draw_arrow(ax, (0,3.5), (0,3))        # BERT → Shared
draw_arrow(ax, (0,2), (-2,1))         # Shared → Sentiment
draw_arrow(ax, (0,2), (2,1))          # Shared → Star Rating

# Title
ax.set_title("Multi-Task BERT Model: Shared Layer → Two Heads", fontsize=14, weight='bold')

plt.show()
```
